# Семинар 3. Регуляризация и градиентное обучение

На Ames сравним МНК, Ridge и Lasso и добавим взаимодействия признаков. На рисунках разберём шаги GD и SGD, затем обновим модель проката велосипедов по мере поступления новых данных.

```python
import numpy as np
import pandas as pd
import sklearn
from IPython.display import display
from sklearn.base import clone
from sklearn.model_selection import train_test_split, KFold, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression, Ridge, Lasso, SGDRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error

SEED = 42

def rmse(y, pred):
    return np.sqrt(mean_squared_error(y, pred))
from sklearn.metrics import r2_score
from sklearn.model_selection import cross_val_score
from sklearn.feature_selection import SelectFromModel
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.dpi': 110, 'font.size': 11, 'axes.grid': True, 'grid.alpha': .2})
```

```python
#@title Загрузить встроенный Ames Housing
import base64, gzip, hashlib, io
_csv = gzip.decompress(base64.b64decode(''))
assert hashlib.sha256(_csv).hexdigest() == '1e18addf81e5e4d347cc17ee6075bbe4a42b7fa26b9e5b063e8f692a5f929d41'
houses = pd.read_csv(io.BytesIO(_csv))
```

## 1. Возвращаемся к модели прошлого занятия

Используем пять числовых признаков и район: площадь, качество дома, год постройки, вместимость гаража, длину фасада участка и `Neighborhood`.

Разбиение прежнее: **train 60%**, **valid 20%**, **test 20%**. Цены теперь в тысячах долларов. На valid сравниваем варианты, test оставляем до конца.

Здесь проверяем прогноз для похожих домов случайным разбиением. Для будущих продаж делили бы по времени, повторные строки одного дома держали бы в одной части.

```python
numeric = ['GrLivArea', 'OverallQual', 'YearBuilt', 'GarageCars', 'LotFrontage']
categorical = ['Neighborhood']
X = houses.set_index('Id')[numeric + categorical]
y = houses.set_index('Id')['SalePrice'] / 1000
X_develop, X_test, y_develop, y_test = train_test_split(X, y, test_size=.2, random_state=SEED)
X_train, X_valid, y_train, y_valid = train_test_split(
    X_develop, y_develop, test_size=.25, random_state=SEED)
display(X_train.join(y_train.rename('Цена, тыс. $')).head())
```

Вспомним обработку: пропуски в числах заменяем медианой, район кодируем one-hot. Добавим `StandardScaler`, чтобы штраф и шаг обучения не зависели от единиц признаков. Вся обработка учится только на train.

```python
prepare = ColumnTransformer([
    ('numbers', Pipeline([
        ('impute', SimpleImputer(strategy='median')),
        ('scale', StandardScaler())
    ]), numeric),
    ('district', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical)
])

linear_model = Pipeline([
    ('prepare', prepare),
    ('model', LinearRegression())
])
linear_model.fit(X_train, y_train)
```

Будем пополнять одну таблицу, как в прошлый раз. Выбираем по **RMSE valid**. Рядом смотрим ошибку train, MAE и **R² valid**. Для ошибок меньше означает лучше, для R² больше.

```python
#@title Таблица сравнения
scores = {}
models = {}

def add_result(name, fitted_model):
    models[name] = fitted_model
    scores[name] = {
        'RMSE train, тыс. $': rmse(y_train, fitted_model.predict(X_train)),
        'RMSE valid, тыс. $': rmse(y_valid, fitted_model.predict(X_valid)),
        'MAE valid, тыс. $': mean_absolute_error(y_valid, fitted_model.predict(X_valid)),
        'R² valid': r2_score(y_valid, fitted_model.predict(X_valid))
    }

def show_results(names=None, cv_scores=None):
    table = pd.DataFrame.from_dict(scores, orient='index')
    if names is not None:
        table = table.loc[names]
    if cv_scores is not None:
        table['RMSE CV, тыс. $'] = cv_scores
    display(table.round(2).assign(**{'R² valid': table['R² valid'].round(3)}))

baseline = DummyRegressor().fit(X_train, y_train)
add_result('Средняя цена', baseline)
add_result('Линейная регрессия', linear_model)
show_results()
```

## 2. Добавляем штраф: Ridge

`Ridge` сдерживает большие веса. Чем больше `alpha`, тем сильнее штраф. Обработку и признаки оставляем прежними, меняем только модель.

**Перед запуском:** что произойдёт с ошибками train и valid при alpha=1 и alpha=100?

```python
for alpha in [1, 100]:
    ridge_model = clone(linear_model)
    ridge_model.set_params(model=Ridge(alpha=alpha))
    ridge_model.fit(X_train, y_train)
    add_result(f'Ridge, alpha={alpha}', ridge_model)
show_results(['Линейная регрессия', 'Ridge, alpha=1', 'Ridge, alpha=100'])
```

**Обсудим:** какой штраф помог, а какой испортил прогноз? Дальше выберем alpha по кросс-валидации.

Посмотрим на веса пяти числовых признаков. Они уже стандартизованы, поэтому вес относится к изменению признака на одно стандартное отклонение.

```python
weights = pd.DataFrame(index=numeric)
for name in ['Линейная регрессия', 'Ridge, alpha=1', 'Ridge, alpha=100']:
    weights[name] = models[name].named_steps['model'].coef_[:len(numeric)]
display(weights.round(2))
```

## 3. Подбираем alpha через GridSearchCV

Вместо ручного перебора проверим сетку на десяти фолдах **внутри train**. Передаём весь Pipeline, чтобы медианы и scaler каждый раз обучались без проверочного фолда.

`model__alpha` означает параметр alpha у шага model. Если лучший вариант окажется на границе сетки, её стоит расширить.

```python
cv = KFold(n_splits=10, shuffle=True, random_state=SEED)
alphas = np.logspace(-2, 3, 16)
search = GridSearchCV(
    clone(linear_model).set_params(model=Ridge()),
    {'model__alpha': alphas},
    scoring='neg_root_mean_squared_error', cv=cv, n_jobs=1,
    return_train_score=True, error_score='raise'
)
search.fit(X_train, y_train)
best_alpha = float(search.best_params_['model__alpha'])
print(f'Лучший alpha по CV: {best_alpha:g}')
add_result(f'Ridge, alpha={best_alpha:g} (CV)', search.best_estimator_)

ols_cv = -cross_val_score(linear_model, X_train, y_train, cv=cv,
                         scoring='neg_root_mean_squared_error')
# alpha=0 означает отсутствие штрафа. Этот случай считаем через МНК.
check_rows = [{'alpha': 0., 'RMSE CV': ols_cv.mean(),
               'RMSE valid': rmse(y_valid, linear_model.predict(X_valid))}]
for a in [.01, .1, 1., best_alpha]:
    j = np.argmin(np.abs(alphas - a))
    fitted = clone(linear_model).set_params(model=Ridge(alpha=alphas[j])).fit(X_train, y_train)
    check_rows.append({'alpha': alphas[j],
                      'RMSE CV': -search.cv_results_['mean_test_score'][j],
                      'RMSE valid': rmse(y_valid, fitted.predict(X_valid))})
display(pd.DataFrame(check_rows).set_index('alpha').round(3))
fig, axes = plt.subplots(1, 2, figsize=(12, 4), layout='constrained')
cv_mean = -search.cv_results_['mean_test_score']
cv_sd = search.cv_results_['std_test_score']
axes[0].errorbar(alphas, cv_mean, yerr=cv_sd, marker='o', capsize=3, label='Ridge: CV ± SD')
axes[0].axhline(ols_cv.mean(), color='black', ls='--', label='МНК: те же фолды')
axes[0].axvline(best_alpha, color='tab:green', ls=':', label=f'Выбрано {best_alpha:g}')
axes[0].set(xscale='log', xlabel='alpha', ylabel='RMSE, тыс. $', title='Как штраф влияет на качество')
axes[0].legend(fontsize=9)

ridge_path = []
for alpha in alphas:
    fitted = clone(linear_model).set_params(model=Ridge(alpha=alpha)).fit(X_train, y_train)
    ridge_path.append(fitted.named_steps['model'].coef_[:len(numeric)])
for j, name in enumerate(numeric):
    axes[1].semilogx(alphas, np.array(ridge_path)[:, j], label=name)
axes[1].set(xlabel='alpha', ylabel='Вес, тыс. $ на одно σ признака', title='Какие веса меняются')
axes[1].legend(fontsize=8)
plt.show()
```

`mean_test_score` относится к проверочным фолдам, а не к отложенному test. Все RMSE здесь в тысячах долларов.

**Нужен ли здесь Ridge?** На текущем valid МНК без штрафа лучше: 33,900 против 34,240 у выбранного Ridge. Малые alpha почти не меняют прогноз. На CV Ridge немного лучше: 36,930 против 37,095. Для этих признаков и разбиения заметной пользы от штрафа нет. Это не вывод о любом варианте модели на Ames.

Усы на графике показывают стандартное отклонение между фолдами, не доверительный интервал. Отдельный вес не обязан сжиматься монотонно.

### Когда нужны дополнительные признаки и штраф

Цена может зависеть от площади нелинейно, а эффект площади может зависеть от качества дома. Добавим квадраты и попарные произведения числовых признаков. Сравним расширенную МНК и Ridge на тех же фолдах.

**Проверяем:** уменьшается ли ошибка на train и сохраняется ли улучшение на новых домах?

```python
from sklearn.preprocessing import PolynomialFeatures

# Масштабируем и исходные числа, и полученные квадраты и произведения.
expanded_prepare = clone(prepare)
expanded_prepare.set_params(numbers=Pipeline([
    ('impute', SimpleImputer(strategy='median')),
    ('scale_before', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('scale_after', StandardScaler())
]))
expanded_ols = Pipeline([('prepare', expanded_prepare), ('model', LinearRegression())])
expanded_ols.fit(X_train, y_train)
add_result('МНК: квадраты и взаимодействия', expanded_ols)

expanded_search = GridSearchCV(
    clone(expanded_ols).set_params(model=Ridge()),
    {'model__alpha': np.logspace(-2, 3, 16)},
    cv=cv, scoring='neg_root_mean_squared_error', error_score='raise'
)
expanded_search.fit(X_train, y_train)
expanded_alpha = expanded_search.best_params_['model__alpha']
add_result(f'Ridge: квадраты и взаимодействия, alpha={expanded_alpha:g}',
           expanded_search.best_estimator_)
expanded_cv = -cross_val_score(expanded_ols, X_train, y_train, cv=cv,
                              scoring='neg_root_mean_squared_error')
show_results(
    ['Линейная регрессия', f'Ridge, alpha={best_alpha:g} (CV)',
     'МНК: квадраты и взаимодействия',
     f'Ridge: квадраты и взаимодействия, alpha={expanded_alpha:g}'],
    cv_scores=[ols_cv.mean(), -search.best_score_,
               expanded_cv.mean(), -expanded_search.best_score_]
)
```

Дополнительные признаки меняют возможности модели, штраф ограничивает веса. Это две разные настройки. Выбираем их по качеству на новых данных, а не по одной ошибке train. При одинаковых признаках GD и SGD меняют способ поиска весов, а не добавляют модели новые зависимости.

В этом запуске дополнительные признаки снижают RMSE valid с **33,90 до 30,54 тыс. $**. Ridge на расширенных признаках даёт **30,88**, зато немного лучше по CV: **32,45 против 32,76**. Главный выигрыш здесь дали признаки. Штраф проверили отдельно и не получили дополнительного улучшения на valid.

## 4. Lasso: качество или меньше признаков?

Подбор по RMSE не требует обнулить много весов. Сравним лучший alpha по CV с **alpha=1**, при котором штраф сильнее. Считаем признаки после one-hot: каждый район здесь отдельный столбец.

**Перед запуском:** сколько качества готовы отдать за более короткую модель?

```python
lasso_search = GridSearchCV(
    clone(linear_model).set_params(model=Lasso(max_iter=30000)),
    {'model__alpha': np.logspace(-2, 1, 7)},
    scoring='neg_root_mean_squared_error', cv=cv, n_jobs=1
)
lasso_search.fit(X_train, y_train)
lasso_model = lasso_search.best_estimator_
lasso_alpha = float(lasso_search.best_params_['model__alpha'])
print(f'Выбрано alpha={lasso_alpha:g}')
add_result(f'Lasso, alpha={lasso_alpha:g} (CV)', lasso_model)

compact_alpha = 1.0
compact_lasso = clone(linear_model).set_params(model=Lasso(alpha=compact_alpha, max_iter=30000))
compact_lasso.fit(X_train, y_train)
add_result('Lasso, alpha=1 (сильнее отбор)', compact_lasso)

lasso_summary = []
for label, a, fitted in [('Минимум RMSE CV', lasso_alpha, lasso_model),
                         ('Сильнее отбор', compact_alpha, compact_lasso)]:
    j = np.argmin(np.abs(np.array(lasso_search.cv_results_['param_model__alpha'], dtype=float) - a))
    weights_lasso = fitted.named_steps['model'].coef_
    lasso_summary.append({'Вариант': label, 'alpha': a,
                         'Осталось весов': np.sum(np.abs(weights_lasso) > 1e-8),
                         'RMSE CV, тыс. $': -lasso_search.cv_results_['mean_test_score'][j],
                         'RMSE valid, тыс. $': rmse(y_valid, fitted.predict(X_valid)),
                         'R² valid': r2_score(y_valid, fitted.predict(X_valid))})
display(pd.DataFrame(lasso_summary).set_index('Вариант').round(3))
```

### После сильного отбора переобучаем МНК

Lasso с alpha=1 оставляет 6 из 29 весов, но ухудшает RMSE valid до 37,01 против 34,13 у лучшего Lasso по CV. Проверим, поможет ли заново оценить веса этих шести признаков без штрафа. Отбор и обработка учатся только на train.

```python
selected_model = Pipeline([
    ('prepare', clone(prepare)),
    ('select', SelectFromModel(Lasso(alpha=compact_alpha, max_iter=30000), threshold=1e-8)),
    ('model', LinearRegression())
])
selected_model.fit(X_train, y_train)
print('Оставлено признаков:', selected_model.named_steps['select'].get_support().sum())
add_result(f'МНК после отбора Lasso, alpha={compact_alpha:g}', selected_model)

show_results(['Линейная регрессия', f'Lasso, alpha={lasso_alpha:g} (CV)',
              'Lasso, alpha=1 (сильнее отбор)',
              f'МНК после отбора Lasso, alpha={compact_alpha:g}'])

fig, axes = plt.subplots(1, 2, figsize=(12, 4), layout='constrained')
lasso_alphas = np.array(lasso_search.cv_results_['param_model__alpha'], dtype=float)
axes[0].semilogx(lasso_alphas, -lasso_search.cv_results_['mean_test_score'], 'o-')
axes[0].axvline(lasso_alpha, color='tab:green', ls=':', label=f'Выбрано {lasso_alpha:g}')
axes[0].set(xlabel='Lasso alpha', ylabel='CV RMSE, тыс. $', title='Выбираем штраф по CV')
axes[0].axvline(compact_alpha, color='tab:orange', ls='--', label='Сильнее отбор: alpha=1')
axes[0].legend()
counts = []
for a in lasso_alphas:
    fitted = clone(linear_model).set_params(model=Lasso(alpha=a, max_iter=30000)).fit(X_train, y_train)
    counts.append(np.sum(np.abs(fitted.named_steps['model'].coef_) > 1e-8))
axes[1].semilogx(lasso_alphas, counts, 'o-')
axes[1].set(xlabel='Lasso alpha', ylabel='Ненулевых коэффициентов', title='Сколько признаков остаётся на train')
plt.show()
```

**Обсудим:** что потеряли при сокращении модели? Помогло ли переобучение без штрафа?

Lasso позволяет получить больше нулей, если усилить штраф. Но число нулей само по себе не мера качества. При связанных признаках Lasso может оставить один из нескольких похожих.

## 5. Почему SGD может быть выгоден: посмотрим на шаги

Вернём эксперимент HSE с двумя весами. Искусственные данные нужны только для рисунка: видно, куда идёт каждый метод. Полный GD берёт все объекты, mini-batch 10, SGD один.

**Меняйте `eta`, `updates` и `decay`.** При большом шаге SGD шумит, уменьшение шага помогает успокоиться около минимума.

```python
rng = np.random.default_rng(100)
X_demo = rng.uniform(-5, 5, (300, 2)) * [1, 3]
y_demo = X_demo @ np.array([1.5, -.7]) + rng.normal(0, 1, 300)
X_demo = StandardScaler().fit_transform(X_demo)
y_demo = y_demo - y_demo.mean()
w_opt = np.linalg.lstsq(X_demo, y_demo, rcond=None)[0]

eta = .1
updates = 150
decay = False

def descent(X, y, batch_size, eta, updates, decay=False):
    rng = np.random.default_rng(SEED)
    w = np.zeros(X.shape[1])
    path = [w.copy()]
    for k in range(updates):
        ids = np.arange(len(y)) if batch_size == len(y) else rng.integers(len(y), size=batch_size)
        grad = 2 * X[ids].T @ (X[ids] @ w - y[ids]) / len(ids)
        step = eta / (1 + k / 30) if decay else eta
        w -= step * grad
        path.append(w.copy())
    return np.array(path)

runs = {name: (b, descent(X_demo, y_demo, b, eta, updates, decay))
        for name, b in [('Full GD', 300), ('Mini-batch', 10), ('SGD', 1)]}
```

```python
#@title Траектории на линиях одинаковой MSE
all_points = np.vstack([path for b, path in runs.values()] + [w_opt[None, :]])
center = (all_points.min(axis=0) + all_points.max(axis=0)) / 2
radius = np.max(np.ptp(all_points, axis=0)) * .6
u = np.linspace(center[0] - radius, center[0] + radius, 150)
v = np.linspace(center[1] - radius, center[1] + radius, 150)
U, V = np.meshgrid(u, v)
H = X_demo.T @ X_demo / len(y_demo)
q = X_demo.T @ y_demo / len(y_demo)
loss_grid = H[0, 0]*U**2 + 2*H[0, 1]*U*V + H[1, 1]*V**2 - 2*q[0]*U - 2*q[1]*V + np.mean(y_demo**2)
colors = ['tab:blue', 'tab:orange', 'tab:green']
fig, axes = plt.subplots(1, 3, figsize=(13, 4.5), sharex=True, sharey=True, layout='constrained')
for ax, (name, (b, path)), color in zip(axes, runs.items(), colors):
    ax.contour(U, V, loss_grid, levels=15, colors='0.75', linewidths=.7)
    ax.plot(path[:, 0], path[:, 1], '.-', color=color, markersize=3, label='Шаги')
    ax.scatter(*path[0], marker='s', color='black', label='Старт')
    ax.scatter(*w_opt, marker='*', s=110, color='red', label='Минимум МНК', zorder=4)
    ax.set(title=f'{name}: {b} ' + ('объект' if b == 1 else 'объектов') + ' на шаг', xlabel='Вес w₁', ylabel='Вес w₂', aspect='equal')
    ax.legend(fontsize=8)
plt.show()
```

```python
# Слева сравниваем обновления, справа учитываем стоимость обновления.
fig, axes = plt.subplots(1, 2, figsize=(12, 4), layout='constrained')
for (name, (b, path)), color in zip(runs.items(), colors):
    loss = np.mean((X_demo @ path.T - y_demo[:, None])**2, axis=0)
    k = np.arange(len(path))
    axes[0].semilogy(k, loss, color=color, label=name)
    axes[1].loglog((k*b)[1:], loss[1:], color=color, label=name)
q_opt = np.mean((X_demo @ w_opt - y_demo)**2)
for ax in axes:
    ax.axhline(q_opt, color='black', ls='--', label='Минимум МНК')
    ax.set_ylabel('MSE, лог. шкала')
    ax.legend(fontsize=9)
axes[0].set(xlabel='Число обновлений', title='Насколько точен один шаг?')
axes[1].set(xlabel='Объектов в вычислениях градиента, лог. шкала', title='Сколько данных обработано ради этих шагов?')
plt.show()
```

**Что получили:** GD делает точные, но дорогие шаги. SGD может быстро приблизиться к хорошим весам по числу обработанных объектов, но около минимума колеблется. Mini-batch уменьшает шум. Это компромисс между стоимостью шага и его точностью, а не способ обязательно получить меньшую ошибку.

Справа показан счётчик работы, не секунды. Расчёт MSE для рисунка в него не входит. Объекты выбираются с возвращением, поэтому 300 выборов здесь не означают полный проход без повторов.

```python
# Постоянный и убывающий шаг на тех же случайных объектах.
fig, ax = plt.subplots(figsize=(9, 3.5), layout='constrained')
for decreasing in [False, True]:
    path = descent(X_demo, y_demo, 1, eta, 400, decay=decreasing)
    loss = np.mean((X_demo @ path.T - y_demo[:, None])**2, axis=0)
    ax.semilogy(loss, label='Шаг уменьшается' if decreasing else 'Постоянный шаг')
ax.axhline(q_opt, color='black', ls='--', label='Минимум МНК')
ax.set(xlabel='Обновления SGD', ylabel='MSE, лог. шкала', title='Снижаем шум около минимума')
ax.legend()
plt.show()
```

### Дополнительно: тот же Ridge через SAG

SAG хранит последние градиенты объектов. В sklearn его можно выбрать как solver у Ridge. Проверим тот же alpha: меняем способ обучения, не критерий. На этом размере данных выигрыш по времени не гарантирован.

## 6. Новые данные приходят порциями: partial_fit

[UCI Bike Sharing](https://archive.ics.uci.edu/dataset/275/bike+sharing+dataset): число аренд за час. Учимся на 2011 году. В 2012 году каждый месяц **сначала прогнозируем, затем обновляем модель**. Сравниваем с её копией, которую больше не обучали.

Обработку и масштаб цели фиксируем по 2011 году. `casual` и `registered` не берём: их сумма равна `cnt`. Погода здесь наблюдаемая, поэтому это оценка спроса при известных условиях, а не прогноз погоды на месяц вперёд.

```python
from pathlib import Path
from urllib.request import urlretrieve
from zipfile import ZipFile
from copy import deepcopy

archive = Path('bike-sharing.zip')
if not archive.exists():
    urlretrieve('https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip', archive)
with ZipFile(archive) as z:
    bikes = pd.read_csv(z.open('hour.csv')).sort_values(['dteday', 'hr'])

bike_cat = ['hr', 'weekday', 'weathersit', 'workingday', 'holiday']
bike_num = ['temp', 'hum', 'windspeed']
initial = bikes[bikes.yr == 0]  # 2011 год
bike_prepare = ColumnTransformer([
    ('categories', OneHotEncoder(handle_unknown='ignore', sparse_output=False), bike_cat),
    ('numbers', StandardScaler(), bike_num)
])
X_initial = bike_prepare.fit_transform(initial[bike_cat + bike_num])
target_scale = StandardScaler()
y_initial = target_scale.fit_transform(initial[['cnt']]).ravel()

updating = SGDRegressor(alpha=0.0001, learning_rate='constant', eta0=0.003,
                       max_iter=10, tol=None, shuffle=True, random_state=SEED)
updating.fit(X_initial, y_initial)  # ровно 10 проходов по 2011 году
fixed = deepcopy(updating)  # clone не подходит: он не сохраняет обученные веса
print(f'{len(bikes)} часов, из них {len(initial)} в начальном обучении')
```

```python
monthly = []
for month, batch in bikes[bikes.yr == 1].groupby('mnth', sort=True):
    X_month = bike_prepare.transform(batch[bike_cat + bike_num])
    y_month = batch.cnt.to_numpy()
    # В прогнозах нет информации о числе аренд текущего месяца.
    pred_fixed = target_scale.inverse_transform(fixed.predict(X_month)[:, None]).ravel()
    pred_updating = target_scale.inverse_transform(updating.predict(X_month)[:, None]).ravel()
    monthly.append({'Месяц': month,
                    'Без обновлений': rmse(y_month, pred_fixed),
                    'С обновлениями': rmse(y_month, pred_updating)})
    # Ответы стали известны: один проход по новой порции, без старых данных.
    updating.partial_fit(X_month, target_scale.transform(batch[['cnt']]).ravel())

monthly = pd.DataFrame(monthly).set_index('Месяц')
ax = monthly.plot(marker='o', style=['--', '-'], figsize=(10, 4))
ax.set(xlabel='Месяц 2012 года', ylabel='RMSE, аренд за час',
       title='Помогает ли обновление на прошедшем месяце?', xticks=range(1, 13))
plt.tight_layout()
plt.show()
print('Средняя месячная RMSE:')
print(monthly.mean().round(1))
wins = (monthly['С обновлениями'] < monthly['Без обновлений']).sum()
print(f'Обновляемая модель лучше в {wins} из 12 месяцев')
```

В этом запуске обновление помогает в **11 из 12 месяцев**. Средняя месячная RMSE **121,3 против 160,3** без обновлений. В январе результаты совпадают: модели ещё одинаковы.

Начиная с февраля обновляемая модель использует ответы за предыдущие месяцы. Сравниваем способность адаптироваться к новым данным, а не два способа решить одну неизменную задачу.

`partial_fit` позволяет доучиваться без повторной обработки всей истории. Это не доказательство, что SGD здесь необходим или быстрее полного переобучения. Средняя месячная RMSE даёт одинаковый вес каждому месяцу.
